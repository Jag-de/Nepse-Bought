```python
debenture_ = Remaining[Remaining['Security Name'].str.contains('debenture|bond', case=False)]
debenture_
```

```python
ordinary_S= Remaining[~Remaining['Security Name'].str.contains('debenture|bond', case=False)]
ordinary_S
```

```python
ordinary_S.shape
```

```python
shl_=ordinary_S[ordinary_S['Security Name'].str.contains('Soaltee ', case=False)]
shl_
```

```python
net_ordinary_S=ordinary_S[~ordinary_S['Security Name'].str.contains('Soaltee ', case=False)]
net_ordinary_S
```

```python
mutual_=mutual_f[['Symbol','Close Price']]
mutual_
```

```python
ordinary_=net_ordinary_S[['Symbol','Close Price']]
ordinary_
```

```python
def SubCategory(debenture_,mutual_,ordinary_,shl_):
    i,j,k,l = 0,0,0,0
    for index, row in debenture_.iterrows():
        i += row['Close Price'] * 25
    for index, row in mutual_.iterrows():
        j += row['Close Price'] * 100
    for index, row in ordinary_.iterrows():
        k += row['Close Price'] * 10 
    for index, row in shl_.iterrows():
        l += row['Close Price'] * 100 
    
    return i,j,k,l
```

```python
SubCategory(debenture_,mutual_,ordinary_,shl_)
```

```python
net_total=sum(SubCategory(debenture_,mutual_,ordinary_,shl_))
```
